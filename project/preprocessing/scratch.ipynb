```python
import pandas as pd

df_pred_speed = pd.read_feather('../data/predictions/speed_pred_test.feather')
df_pred_current = pd.read_feather('../data/predictions/current_pred_test.feather')
df_pred_temp = pd.read_feather('../data/predictions/temp_pred_test.feather')

df_pred_current.drop(columns=['time'], inplace=True)
df_pred_temp.drop(columns=['time'], inplace=True)

df_merged = pd.concat([df_pred_speed.reset_index(drop=True),
                       df_pred_current.reset_index(drop=True)],
                      axis=1)

df_merged = pd.concat([df_merged.reset_index(drop=True),
                       df_pred_temp.reset_index(drop=True)],
                      axis=1)

df_merged.head()
```
